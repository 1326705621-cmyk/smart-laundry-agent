## 6. 关键概念

- **Conda 环境**：项目专用的 Python 和依赖空间，不影响 `base` 或 PyTorch 环境。
- **依赖**：项目运行所需的第三方工具，例如 Streamlit。
- **环境变量**：把路径、城市和密钥放在代码外部的配置方式。
- **入口文件**：启动应用时首先执行的文件，本项目是 `app.py`。
- **测试**：用代码自动检查另一段代码是否符合预期。

## 7. 为什么这样设计

本项目选择 `src/smart_laundry` 保存业务代码，而让 `app.py` 只装配页面。代价是初期多了几个目录，但后续数据库、天气和 Agent 可以分别测试，不会形成难以维护的超大页面文件。依赖统一写在 `pyproject.toml`，避免两份依赖清单互相不一致。

## 8. 常见错误与排查

1. **提示找不到 Python**：先在 Anaconda Prompt 运行 `conda activate smart-laundry`。
2. **提示找不到 streamlit**：确认环境已激活，再运行 `python -m pip install -e ".[dev]"`。
3. **导入 smart_laundry 失败**：确认安装命令是在项目根目录执行。
4. **页面端口被占用**：关闭旧页面进程，或让 Streamlit 使用另一个端口。
5. **密钥泄露风险**：真实值只写入 `.env`，不要修改 `.env.example` 放入真实密钥。

## 9. 面试可能追问

**问：为什么不用 Anaconda 的 base 环境？**  答：独立环境隔离依赖版本，减少不同项目互相影响，也便于复现。继续追问可能是：如何导出依赖、如何删除或重建环境。

**问：为什么配置不直接写在 app.py？**  答：集中配置便于测试和替换运行环境，也避免业务代码到处接触密钥。继续追问可能是：生产环境如何注入密钥。

**问：为什么使用 Streamlit？**  答：它能用 Python 快速形成可演示网页，适合本项目先验证业务闭环；复杂前端交互不是首版重点。
