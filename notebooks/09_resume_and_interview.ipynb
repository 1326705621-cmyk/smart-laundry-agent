## 5. 数据流
本地源代码 → `.gitignore` 排除私人文件 → 自动测试和敏感信息扫描 → Git 首次提交 → GitHub 公开仓库 → 招聘者按 README 安装、运行并查看文档。在线 Demo 是单独的部署步骤，不等同于 GitHub 仓库。

## 6. 关键概念
- **Repository**：包含代码、文档和历史提交的仓库。
- **Commit**：一次有说明的代码快照。
- **Remote**：GitHub 上与本地仓库对应的远程地址。
- **License**：说明他人可以怎样使用项目。
- **可复现**：别人能按文档在新环境中运行项目。

## 7. 为什么这样设计
简历强调 Tool Calling、可靠性和数据闭环，因为这些是已经实现且能通过代码和测试解释的能力；不声称生产高并发或真实用户量，因为项目还没有相应部署与数据。

## 8. 常见错误与排查
- `git add .` 带入数据库：先执行 `git status --short --untracked-files=all` 和 `git check-ignore`。
- README 只能在作者电脑运行：不要写固定的 `C:\aimini` 路径。
- GitHub 有代码但没有网页：仓库链接和在线 Demo 是两件事。
- 简历测试数量过期：新增测试后同步更新 README、简历和 Notebook。
